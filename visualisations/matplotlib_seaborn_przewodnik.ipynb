### Dane przykładowe: zbiór „sprzedaż", zawiera wszystkie typy danych

Kolumny: data/czas, kategoria nominalna (`region`, `channel`), kategoria uporządkowana (`segment`), liczba całkowita (`customer_age`, `units`), ciągłe (`income`, `unit_price`, `discount`, `revenue`, `delivery_days`), porządkowa 1–5 (`satisfaction`), logiczna (`is_returning`), tekst/ID (`order_id`), braki danych (`delivery_days`).

````python
rng = np.random.default_rng(42)
n = 1500

df = pd.DataFrame(
    {
        "order_id": [f"ORD-{i:05d}" for i in range(n)],
        "order_date": pd.Timestamp("2023-01-01")
        + pd.to_timedelta(rng.integers(0, 730, n), unit="D"),
        "region": rng.choice(
            ["Północ", "Południe", "Wschód", "Zachód", "Centrum"],
            n,
            p=[0.2, 0.15, 0.2, 0.2, 0.25],
        ),
        "channel": rng.choice(["Online", "Sklep", "Telefon"], n, p=[0.55, 0.35, 0.10]),
        "segment": pd.Categorical(
            rng.choice(["Basic", "Standard", "Premium"], n, p=[0.45, 0.40, 0.15]),
            categories=["Basic", "Standard", "Premium"],
            ordered=True,
        ),
        "customer_age": np.clip(rng.normal(40, 12, n), 18, 80).round().astype(int),
        "is_returning": rng.random(n) < 0.4,
    }
)

seg_mult = (
    df["segment"]
    .astype(str)
    .map({"Basic": 1.0, "Standard": 1.6, "Premium": 2.8})
    .to_numpy()
)
df["income"] = (
    rng.lognormal(8.2, 0.35, n) * (1 + (df["customer_age"] - 18) / 120)
).round(0)
df["units"] = rng.poisson(3, n) + 1
df["unit_price"] = (rng.gamma(6, 12, n) * seg_mult).round(2)
df["discount"] = np.clip(rng.beta(2, 12, n), 0, 0.5).round(3)
df["revenue"] = (df["units"] * df["unit_price"] * (1 - df["discount"])).round(2)
df["delivery_days"] = (
    rng.gamma(3, 1.2, n) + (df["channel"] == "Online").to_numpy() * 1.5
).round(1)
df["satisfaction"] = np.clip(
    np.round(
        4.2
        - 0.15 * df["delivery_days"]
        + 0.3 * (df["segment"] == "Premium").to_numpy()
        + rng.normal(0, 0.8, n)
    ),
    1,
    5,
).astype(int)

# braki danych i kolumny pochodne
df.loc[rng.choice(n, 40, replace=False), "delivery_days"] = np.nan
df["month"] = df["order_date"].dt.to_period("M").dt.to_timestamp()
df["year"] = df["order_date"].dt.year
df["weekday"] = df["order_date"].dt.day_name()
df["returning_int"] = df["is_returning"].astype(int)  # 0/1 do regresji logistycznej

# agregat miesięczny (do wykresów liniowych)
df_month = df.groupby(["month", "channel"], as_index=False).agg(
    revenue=("revenue", "sum"), orders=("order_id", "size")
)

# dane „szerokie" (do heatmap)
corr = df.select_dtypes("number").drop(columns=["year", "returning_int"]).corr()
pivot = df.pivot_table(
    index="region", columns="segment", values="revenue", aggfunc="mean", observed=True
)

smp = df.sample(
    600, random_state=4
)  # próbka do wykresów punktowych (czytelniejsze, szybsze)

df.head()
````

## 1. Matplotlib: anatomia i formatowanie

**Model obiektowy:** `Figure` (całe płótno) → `Axes` (pojedynczy wykres: osie, dane, tytuł) → elementy (`Line2D`, `Patch`, `Text`, `Axis`). Zawsze używaj stylu obiektowego (`fig, ax = plt.subplots()`), a nie `plt.plot()`. Daje kontrolę i działa przy wielu osiach.

**Dwa rodzaje funkcji seaborn:**

| Typ | Przykłady | Cechy |
|---|---|---|
| *axes-level* | `scatterplot`, `lineplot`, `histplot`, `kdeplot`, `boxplot`, `barplot`, `heatmap`… | Rysują na istniejącym `ax=`, zwracają `Axes`, łatwo łączyć w subplots |
| *figure-level* | `relplot`, `displot`, `catplot`, `lmplot`, `jointplot`, `pairplot`, `clustermap` | Tworzą własną figurę, mają `col=`/`row=`, zwracają `FacetGrid`/`JointGrid`; **nie przyjmują `ax=`**, rozmiar przez `height=` i `aspect=` |

### 1.0. Najprostsze wykresy w czystym matplotlib

Zanim przejdziesz do formatowania: minimalna składnia podstawowych wykresów. Najprostszy wariant to samo `plt.plot(y)` (matplotlib sam utworzy `Figure` i `Axes`).

| Funkcja | Wymagane ✅ | Częste opcjonalne ⬜ |
|---|---|---|
| `ax.plot(x, y)` | `y` (bez `x` użyje 0, 1, 2…) | `color`, `linestyle`, `linewidth`, `marker`, `label` |
| `ax.bar(x, height)` | `x`, `height` | `color`, `width`, `edgecolor`, `label` (`ax.barh` = poziome) |
| `ax.hist(x)` | `x` | `bins`, `color`, `density`, `alpha`, `cumulative` |
| `ax.scatter(x, y)` | `x`, `y` | `s` (rozmiar), `c` (kolor/wartości), `alpha`, `cmap`, `marker` |
| `ax.boxplot(data)` | `data` (lista tablic) | `vert`, `notch`, `showmeans`, `whis` |
| `ax.pie(x)` | `x` | `labels`, `autopct`, `startangle`, `explode`, `colors` |
| `ax.imshow(matrix)` | `matrix` | `cmap`, `vmin`, `vmax`, `aspect` + `plt.colorbar()` |

**Pozycjonowanie legendy poza wykresem:**

#### Praktyczne sterowanie osiami: ticki, zakresy, skale i oś pomocnicza

- `set_xlim()` / `set_ylim()` ograniczają zakres widoczny na wykresie.
- `set_xticks()` / `set_yticks()` ustawiają pozycje ticków; etykiety najlepiej przekazywać w tym samym wywołaniu.
- `tick_params()` formatuje znaczniki i etykiety osi.
- `set_xscale()` / `set_yscale()` zmieniają skalę, np. na logarytmiczną.
- `secondary_xaxis()` / `secondary_yaxis()` dodają drugą skalę tej samej wielkości (np. m i km). Do innej miary, np. PLN i liczby zamówień, służy `twinx()` – omówiony dalej.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.2), constrained_layout=True)

# a) Ticki, zakres i konwersja jednostek: metry -> kilometry
x_m = np.linspace(0, 10_000, 80)
y = np.sqrt(x_m)
axes[0].plot(x_m, y, color="#4c72b0", linewidth=2)
axes[0].set_xlim(0, 10_000)
axes[0].set_ylim(0, 110)
axes[0].set_xticks([0, 2_500, 5_000, 7_500, 10_000])
axes[0].tick_params(axis="both", which="major", labelsize=9, length=5, width=1)
axes[0].grid(axis="y", alpha=0.25)
axes[0].set(title="Ticki i zakres osi", xlabel="Odległość [m]", ylabel="Wartość")
secondary = axes[0].secondary_xaxis(
    "top", functions=(lambda meters: meters / 1_000, lambda km: km * 1_000)
)
secondary.set_xlabel("Odległość [km]")

# b) Skala logarytmiczna – używaj, gdy wartości różnią się rzędami wielkości
x = np.arange(1, 8)
values = np.array([2, 8, 25, 90, 350, 1_400, 5_500])
axes[1].plot(x, values, marker="o", color="#dd8452")
axes[1].set_yscale("log")
axes[1].set_xticks(x, [f"Grupa {i}" for i in x], rotation=25, ha="right")
axes[1].set(title="Oś Y w skali logarytmicznej", xlabel="Grupa", ylabel="Wartość (log)")
axes[1].grid(axis="y", which="both", alpha=0.25)
plt.show()
````

**Zasada stosowana w całym przewodniku: legenda leży poza obszarem wykresu**, żeby nigdy nie zasłaniała danych.

| Sytuacja | Rozwiązanie |
|---|---|
| Legenda utworzona przez seaborn (`hue=`, `style=`, `size=`) w `ax` | `sns.move_legend(ax, "upper left", bbox_to_anchor=(1.01, 1))` (`title=`, `frameon=`, `ncol=` ✅ opcjonalne) |
| Legenda matplotlib (`label=` + `ax.legend()`) | `ax.legend(loc="upper left", bbox_to_anchor=(1.01, 1), borderaxespad=0)` |
| Legenda pozioma nad wykresem | `ax.legend(loc="lower left", bbox_to_anchor=(0, 1.02), ncol=3, frameon=False)` |
| Wykresy figure-level (`relplot`, `catplot`, `lmplot`, `pairplot`) | Legenda jest już poza wykresem; zmiana: `sns.move_legend(g, "upper center", bbox_to_anchor=(0.5, 1.05), ncol=3)` |
| Ukrycie legendy (gdy `hue` = oś X) | `legend=False` w funkcji seaborn |

W układach z `constrained_layout=True` miejsce na legendę poza osiami jest rezerwowane automatycznie. Przy `tight_layout()` i zapisie użyj `bbox_inches="tight"`.

### 1.5. Łączenie różnych wykresów i kategorii na jednej osi

Są dwa częste przypadki:

### Ważne: co oznaczają wąsy w `barplot`?

Domyślne `errorbar=("ci", 95)` oznacza 95-procentowy przedział ufności dla estymowanej statystyki (domyślnie średniej). **Nie jest to zakres, w którym znajduje się 95% pojedynczych obserwacji.**

| Ustawienie | Co pokazuje | Kiedy używać |
|---|---|---|
| `errorbar=("ci", 95)` | Przedział ufności estymatora | Niepewność oszacowania średniej/medianowej statystyki |
| `errorbar="se"` | Błąd standardowy | Skrótowa prezentacja niepewności estymatora |
| `errorbar="sd"` | Odchylenie standardowe | Zmienność obserwacji wokół średniej |
| `errorbar=("pi", 95)` | Przedział percentylowy | Zakres obejmujący określony odsetek obserwacji |
| `errorbar=None` | Bez przedziału | Gdy prezentujesz wcześniej policzoną wartość lub nie chcesz pokazywać niepewności |

Nie interpretuj nakładających się lub nienakładających się przedziałów jako formalnego testu istotności różnicy. Do wnioskowania potrzebna jest odpowiednia procedura statystyczna.

**Ważne:** jeżeli celem jest pokazanie sumy, zwykle lepiej najpierw zagregować dane (`groupby`) i narysować wynik przez `barplot`/`bar`, zamiast traktować domyślną estymację jako gotową analizę.

### 4.1. `sns.barplot`: parametry

| Parametr | Opis |
|---|---|
| `data`, `x`, `y` ✅ | Dane i zmienne (kategoria + wartość; zamień osie → słupki poziome) |
| `hue` ⬜ | Podział na podgrupy (słupki grupowane); `dodge=False` łączy je w jeden słupek |
| `estimator` ⬜ | `"mean"` (domyślnie), `"sum"`, `"median"`, `"count"`, funkcja |
| `errorbar` ⬜ | `("ci", 95)`, `"sd"`, `"se"`, `None` |
| `capsize` ⬜ | Szerokość kresek na końcach wąsów |
| `order`, `hue_order` ⬜ | Kolejność kategorii |
| `orient` ⬜ | `"v"`/`"h"` (zwykle wykrywane) |
| `width` ⬜ | Szerokość słupka (0–1) |
| `gap` ⬜ | Odstęp między słupkami w grupie (0.13+) |
| `saturation` ⬜ | Nasycenie kolorów (1 = pełne) |
| `fill` ⬜ | `False` = same obrysy |
| `palette`, `color` ⬜ | Paleta (dla `hue`) lub jeden kolor |
| `err_kws` ⬜ | np. `{"linewidth": 1.5, "color": "black"}` |
| `ax` ⬜ | Osie |
| `**kwargs` ⬜ | `edgecolor`, `linewidth`, `alpha`, `hatch` |

````python
fig, axes = plt.subplots(2, 3, figsize=(16, 9), constrained_layout=True)

# a) średnia + 95% CI, wyraźna kolejność
order = df.groupby("region")["revenue"].mean().sort_values(ascending=False).index
sns.barplot(
    df,
    x="region",
    y="revenue",
    order=order,
    hue="region",
    legend=False,
    palette="viridis",
    errorbar=("ci", 95),
    capsize=0.15,
    err_kws={"linewidth": 1.5},
    ax=axes[0, 0],
)
axes[0, 0].set_title("Średnia + 95% CI (posortowane)")

# b) suma zamiast średniej, bez wąsów, etykiety wartości
sns.barplot(
    df,
    x="region",
    y="revenue",
    estimator="sum",
    errorbar=None,
    order=order,
    color="#4c72b0",
    ax=axes[0, 1],
)
axes[0, 1].bar_label(axes[0, 1].containers[0], fmt="{:,.0f}", padding=3, fontsize=9)
axes[0, 1].set_title("Suma + etykiety (bar_label)")
axes[0, 1].yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda v, _: f"{v / 1e3:,.0f}k")
)

# c) poziome słupki, hue (grupowane), odstęp w grupie
sns.barplot(
    df,
    y="region",
    x="revenue",
    hue="segment",
    estimator="mean",
    errorbar=None,
    order=order,
    palette="rocket",
    gap=0.1,
    width=0.8,
    edgecolor="white",
    ax=axes[0, 2],
)
axes[0, 2].set_title("Poziome, grupowane (hue)")
sns.move_legend(axes[0, 2], "upper left", bbox_to_anchor=(1.01, 1), title="Segment")

# d) wyróżnienie jednego słupka (słownik kolorów)
tot = (
    df.groupby("region", as_index=False)["revenue"]
    .sum()
    .sort_values("revenue", ascending=False)
)
cmap_hl = {
    r: ("#d62728" if r == tot.iloc[0]["region"] else "#b8c0cc") for r in tot["region"]
}
sns.barplot(
    tot,
    x="region",
    y="revenue",
    hue="region",
    palette=cmap_hl,
    legend=False,
    ax=axes[1, 0],
)
axes[1, 0].bar_label(axes[1, 0].containers[0], fmt="{:,.0f}", padding=2)
axes[1, 0].axhline(tot["revenue"].mean(), color="black", linestyle="--", linewidth=1)
axes[1, 0].text(
    4.4, tot["revenue"].mean(), " średnia", va="bottom", ha="right", fontsize=9
)
axes[1, 0].set_title("Wyróżniony najlepszy region")

# e) formatowanie pojedynczych słupków PO narysowaniu (patches)
sns.barplot(tot, x="region", y="revenue", color="#8da0cb", ax=axes[1, 1])
for bar, val in zip(axes[1, 1].patches, tot["revenue"]):
    if val < tot["revenue"].mean():  # słupki poniżej średniej
        bar.set_facecolor("#f4a261")
        bar.set_hatch("//")
        bar.set_edgecolor("black")
axes[1, 1].set_title("Warunkowe formatowanie (patches)")

# f) wypełnienie i hatch jako obrys
sns.barplot(
    tot,
    x="region",
    y="revenue",
    fill=False,
    edgecolor="black",
    linewidth=2,
    hatch="..",
    errorbar=None,
    ax=axes[1, 2],
)
axes[1, 2].set_title("fill=False + hatch")

for ax in axes.flat:
    ax.set_xlabel("")
plt.show()
````

### 4.3. Lollipop i dumbbell chart

- **Lollipop:** jeden punkt i łączący go odcinek od wspólnego punktu odniesienia; uproszczona alternatywa dla słupków.
- **Dumbbell:** dwa punkty połączone odcinkiem dla każdej kategorii; porównuje dwie wartości, np. dwa kanały lub dwa okresy.

Oba wykresy są czytelne przy umiarkowanej liczbie kategorii. Lollipop ma sens głównie wtedy, gdy ranking ma wyraźny punkt odniesienia (często zero).

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.8), constrained_layout=True)

# a) Lollipop: średni przychód na region
region_mean = df.groupby("region")["revenue"].mean().sort_values()
axes[0].hlines(
    y=region_mean.index, xmin=0, xmax=region_mean.values, color="#9aa6b2", linewidth=2
)
axes[0].scatter(region_mean.values, region_mean.index, s=85, color="#4c72b0", zorder=3)
axes[0].set(
    title="Lollipop: średni przychód", xlabel="Średni przychód", ylabel="Region"
)
axes[0].grid(axis="x", alpha=0.25)

# b) Dumbbell: porównanie średniego przychodu dwóch kanałów w każdym regionie
channel_region = (
    df.groupby(["region", "channel"], observed=True)["revenue"]
    .mean()
    .unstack("channel")
)
channel_region = channel_region.sort_values("Online")
for region, row in channel_region.iterrows():
    left, right = row["Online"], row["Sklep"]
    axes[1].plot(
        [left, right], [region, region], color="#aab2bd", linewidth=2, zorder=1
    )
    axes[1].scatter(
        left,
        region,
        color="#4c72b0",
        s=60,
        label="Online" if region == channel_region.index[0] else None,
        zorder=3,
    )
    axes[1].scatter(
        right,
        region,
        color="#dd8452",
        s=60,
        label="Sklep" if region == channel_region.index[0] else None,
        zorder=3,
    )
axes[1].set(
    title="Dumbbell: Online vs Sklep", xlabel="Średni przychód", ylabel="Region"
)
axes[1].grid(axis="x", alpha=0.25)
axes[1].legend(title="Kanał")
plt.show()
````

### 4.4. Dot Plot

Dot plot używa punktów zamiast słupków. Jest oszczędny wizualnie i przydatny do rankingu kategorii, zwłaszcza gdy etykiety są długie. W przeciwieństwie do lollipopa nie rysuje odcinka od wartości bazowej do punktu.

````python
region_total = df.groupby("region")["revenue"].sum().sort_values()
fig, ax = plt.subplots(figsize=(8, 4.5), constrained_layout=True)
ax.scatter(region_total.values, region_total.index, s=85, color="#4c72b0", zorder=3)
for value, label in zip(region_total.values, region_total.index):
    ax.annotate(
        f"{value:,.0f}",
        (value, label),
        xytext=(7, 0),
        textcoords="offset points",
        va="center",
        fontsize=8,
    )
ax.set(title="Dot plot: przychód według regionu", xlabel="Przychód", ylabel="Region")
ax.grid(axis="x", alpha=0.25)
plt.show()
````

### 4.5. Pareto Chart

Pareto łączy słupki posortowane malejąco z linią skumulowanego udziału. Linia 80% jest pomocniczą regułą, a nie prawem – nie należy zakładać, że każdy zbiór danych spełnia zasadę 80/20. Do udziału skumulowanego stosujemy drugą oś Y.

````python
pareto = (
    df.groupby("region", as_index=False)["revenue"]
    .sum()
    .sort_values("revenue", ascending=False)
)
pareto["share_cum"] = pareto["revenue"].cumsum() / pareto["revenue"].sum() * 100

fig, ax = plt.subplots(figsize=(15, 5), constrained_layout=True)
ax.bar(pareto["region"], pareto["revenue"], color="#4c72b0")
ax.set(
    title="Pareto: udział regionów w przychodzie", xlabel="Region", ylabel="Przychód"
)
ax.tick_params(axis="x", rotation=20)
ax.yaxis.set_major_formatter(mticker.StrMethodFormatter("{x:,.0f}"))
ax2 = ax.twinx()
ax2.plot(
    pareto["region"], pareto["share_cum"], color="#c44e52", marker="o", linewidth=2
)
ax2.axhline(80, color="gray", linestyle="--", linewidth=1)

ax2.text(
    0.99,
    80,
    "Próg 80%",
    transform=ax2.get_yaxis_transform(),
    ha="right",
    va="bottom",
    color="gray",
    fontsize=9,
)
ax2.set(ylabel="Skumulowany udział [%]", ylim=(0, 110))
ax2.yaxis.set_major_formatter(mticker.PercentFormatter(xmax=100))
ax2.legend(loc="lower right")
plt.show()
````

### 4.6. Waterfall Chart

Waterfall pokazuje, jak dodatnie i ujemne składniki budują zmianę od wartości początkowej do końcowej. Typowe zastosowania: mostek przychodów, kosztów, marży lub wyniku. W przykładzie pierwszy i ostatni słupek są wartościami całkowitymi, a środkowe pokazują przyrosty.

````python
# Przykładowe składniki mostka; wartości są jawnie zdefiniowane.
changes = pd.Series(
    [120_000, 35_000, -18_000, 12_000, -27_000],
    index=["Start", "Wolumen", "Rabat", "Nowi klienci", "Koszty"],
)
start_value = changes.iloc[0]
deltas = changes.iloc[1:]
running = start_value

fig, ax = plt.subplots(figsize=(10, 5), constrained_layout=True)
ax.bar(0, start_value, color="#4c72b0", width=0.7)
for pos, (label, delta) in enumerate(deltas.items(), start=1):
    bottom = running if delta >= 0 else running + delta
    ax.bar(
        pos,
        abs(delta),
        bottom=bottom,
        color="#55a868" if delta >= 0 else "#c44e52",
        width=0.7,
    )
    ax.plot([pos - 0.35, pos + 0.35], [running, running], color="gray", linewidth=0.9)
    running += delta
ax.bar(len(changes), running, color="#4c72b0", width=0.7)
labels = ["Start", *deltas.index, "Koniec"]
ax.set_xticks(range(len(labels)), labels, rotation=20, ha="right")
ax.set(title="Waterfall: mostek zmiany wyniku", ylabel="Wartość")
ax.yaxis.set_major_formatter(mticker.StrMethodFormatter("{x:,.0f}"))
ax.grid(axis="y", alpha=0.2)
plt.show()
````

### Normalizacja histogramu – `stat`

Wysokość słupka zależy od parametru `stat`. Nie porównuj histogramów z różnymi ustawieniami bez wyjaśnienia skali.

| `stat` | Interpretacja |
|---|---|
| `"count"` | Liczba obserwacji w koszyku |
| `"frequency"` | Liczba obserwacji podzielona przez szerokość koszyka |
| `"probability"` | Udział obserwacji w koszyku; suma udziałów wynosi 1 |
| `"percent"` | Udział procentowy obserwacji w koszyku |
| `"density"` | Gęstość; pole pod histogramem wynosi 1 |

Przy porównywaniu grup o różnych liczebnościach rozważ `stat="density"` oraz `common_norm=False`. Wówczas każda grupa jest normalizowana osobno. Pamiętaj, że zmiana szerokości koszyków wpływa na kształt histogramu.

Przydatne parametry:
- `binwidth` – jawna szerokość koszyka;
- `binrange` – zakres danych uwzględniany w koszykach;
- `bins="fd"` – reguła Freedmana–Diaconisa jako automatyczny wybór liczby koszyków;
- `weights` – wagi obserwacji;
- `common_bins` – wspólne granice koszyków dla grup.

**Interpretacja przykładu:** średnia > mediana ⇒ skośność prawostronna (ogon po prawej). Przy takim rozkładzie `log_scale=True` ujawnia strukturę, a do opisu lepsza jest mediana.

### Boxplot, violinplot i ECDF – który wybrać?

| Wykres | Najlepiej odpowiada na pytanie | Ograniczenie |
|---|---|---|
| Boxplot | Gdzie leży mediana, kwartyle i obserwacje poza wąsami? | Nie pokazuje dokładnego kształtu rozkładu |
| Violinplot | Jaki jest kształt rozkładu i czy widać kilka skupień? | Kształt zależy od estymacji KDE i bandwidth |
| ECDF | Jaki odsetek obserwacji jest mniejszy lub równy danemu progowi? | Mniej intuicyjny dla osób nieznających dystrybuanty |

Punkty poza wąsami boxplota są obserwacjami odległymi według przyjętej reguły (zwykle 1,5 × IQR), a nie automatycznie błędnymi rekordami.

W violinplot nie istnieje uniwersalny próg liczebności gwarantujący wiarygodny kształt. Przy małych próbach KDE może tworzyć pozorną strukturę. Zmieniaj `bw_adjust`, sprawdzaj liczebność grup i porównuj z histogramem/ECDF.

## 7. Wykresy kategoryczne (rozkład w grupach)

### 7.1. Boxplot: `sns.boxplot`

**Kiedy:** porównanie rozkładów wielu grup w skondensowanej formie (mediana, rozrzut, outliery).
**Interpretacja:** linia w pudełku = **mediana**; pudełko = **IQR** (Q1–Q3, środkowe 50% danych); wąsy = ostatnie obserwacje w zakresie `whis × IQR` (domyślnie 1,5); punkty poza wąsami = potencjalne outliery. Przesunięcie mediany względem środka pudełka = skośność. Box **nie pokazuje wielomodalności** i liczebności grup.

#### Wersja podstawowa (wymagane: `data`, `x`, `y`)

````python
sns.boxplot(data=df, x="segment", y="revenue")
plt.show()

ax = sns.boxplot(data=df, x="segment", y="revenue", hue="channel")
sns.move_legend(ax, "upper left", bbox_to_anchor=(1.01, 1))
plt.show()
````

| Parametr | Opis |
|---|---|
| `data`, `x`, `y` ✅ | Kategoria + zmienna ciągła (zamień osie → poziomo) |
| `hue` ⬜ | Podgrupy; `dodge`, `gap` sterują odstępem |
| `order`, `hue_order` ⬜ | Kolejność |
| `whis` ⬜ | Zasięg wąsów: `1.5` (domyślnie), `(5, 95)` = percentyle, `(0, 100)` = min–max |
| `notch` ⬜ | Wcięcie = przybliżony CI mediany (nie nakładające się wcięcia sugerują różnicę median) |
| `showmeans` ⬜ | Pokazuje średnią; `meanprops` styl znacznika |
| `showfliers`, `fliersize` ⬜ | Pokazywanie/rozmiar outlierów (`flierprops` pełny styl) |
| `width` ⬜ | Szerokość pudełka |
| `fill` ⬜ | `False` = same obrysy |
| `saturation`, `linewidth`, `linecolor`, `color`, `palette` ⬜ | Wygląd |
| `boxprops`, `whiskerprops`, `capprops`, `medianprops` ⬜ | Style elementów (dict, przekazywane do matplotlib) |
| `log_scale` ⬜ | Skala log |

````python
fig, axes = plt.subplots(2, 3, figsize=(17, 9), constrained_layout=True)

sns.boxplot(
    df,
    x="segment",
    y="revenue",
    hue="segment",
    palette="viridis",
    legend=False,
    ax=axes[0, 0],
)
axes[0, 0].set_title("Podstawowy")

sns.boxplot(
    df,
    x="segment",
    y="revenue",
    hue="channel",
    palette="Set2",
    gap=0.1,
    width=0.7,
    ax=axes[0, 1],
)
axes[0, 1].set_title("hue (grupowany) + gap")
sns.move_legend(axes[0, 1], "upper left", bbox_to_anchor=(1.01, 1))

sns.boxplot(
    df,
    x="segment",
    y="revenue",
    hue="segment",
    legend=False,
    palette="pastel",
    showmeans=True,
    meanprops={
        "marker": "D",
        "markerfacecolor": "white",
        "markeredgecolor": "black",
        "markersize": 7,
    },
    medianprops={"color": "black", "linewidth": 2},
    whiskerprops={"linewidth": 1.5},
    capprops={"linewidth": 1.5},
    flierprops={
        "marker": "o",
        "markersize": 3,
        "alpha": 0.4,
        "markerfacecolor": "gray",
    },
    ax=axes[0, 2],
)
axes[0, 2].set_title("Pełne formatowanie elementów")

sns.boxplot(
    df,
    y="segment",
    x="revenue",
    hue="segment",
    legend=False,
    palette="mako",
    notch=True,
    whis=(5, 95),
    showfliers=False,
    ax=axes[1, 0],
)
axes[1, 0].set_title("Poziomy, notch, whis=(5,95), bez outlierów")

sns.boxplot(
    df,
    x="segment",
    y="revenue",
    fill=False,
    linewidth=2,
    color="#264653",
    log_scale=True,
    ax=axes[1, 1],
)
axes[1, 1].set_title("fill=False + log_scale")

# Boxplot + punkty (strip) = widać i podsumowanie, i dane
sns.boxplot(
    df,
    x="region",
    y="delivery_days",
    color="white",
    linewidth=1.5,
    showfliers=False,
    ax=axes[1, 2],
)
sns.stripplot(
    df.sample(500, random_state=1),
    x="region",
    y="delivery_days",
    size=3,
    alpha=0.5,
    jitter=0.25,
    color="#e76f51",
    ax=axes[1, 2],
)
axes[1, 2].set_title("Box + strip")
plt.show()
````

### 7.7. `sns.catplot`: figure-level interfejs dla wykresów kategorycznych

`catplot()` **nie jest nowym typem wykresu**. To funkcja figure-level, która opakowuje wykresy kategoryczne Seaborn i dodaje wygodne dzielenie na panele (`row=`, `col=`, `col_wrap=`).

- `sns.boxplot(data=df, x="segment", y="revenue", ax=ax)` – tworzy boxplot na wskazanych, istniejących osiach. Daje pełną kontrolę nad układem i łatwe nakładanie innych warstw.
- `sns.catplot(data=df, x="segment", y="revenue", kind="box")` – rysuje **ten sam boxplot**, ale tworzy własną figurę. Zyskujesz łatwe panele, np. `col="channel"` albo `row="segment"`.
- `kind` wybiera rzeczywisty typ wykresu: `"strip"`, `"swarm"`, `"box"`, `"violin"`, `"boxen"`, `"point"`, `"bar"`, `"count"`.

Dlatego na jednym panelu `catplot(kind="box")` wygląda tak samo jak `boxplot()` – różnica dotyczy poziomu API i organizacji paneli, a nie definicji boxplota.

````python
# 1) Jeden panel – ten sam rodzaj wykresu co sns.boxplot.
g = sns.catplot(
    data=df,
    x="segment",
    y="revenue",
    kind="box",
    height=3.8,
    aspect=1.2,
    showfliers=False,
)
g.set_axis_labels("Segment", "Przychód")
g.fig.suptitle("catplot(kind='box'): pojedynczy boxplot", y=1.03)
plt.show()

# 2) Ten sam boxplot rozdzielony na panele według kanału.
g = sns.catplot(
    data=df,
    x="segment",
    y="revenue",
    kind="box",
    col="channel",
    height=3.5,
    aspect=0.95,
    sharey=True,
    showfliers=False,
)
g.set_axis_labels("Segment", "Przychód")
g.set_titles("Kanał: {col_name}")
g.tight_layout()
plt.show()
````

````python
# Możesz też kodować podgrupę kolorem (hue), zachowując panele według kanału.
g = sns.catplot(
    data=df,
    x="segment",
    y="revenue",
    hue="is_returning",
    col="channel",
    kind="box",
    height=3.5,
    aspect=0.95,
    palette={True: "#2a9d8f", False: "#e76f51"},
    showfliers=False,
    sharey=True,
)
g.set_axis_labels("Segment", "Przychód")
g.set_titles("Kanał: {col_name}")
g.add_legend(title="Powracający")
g.tight_layout()
plt.show()
````

### Podsumowanie wykresów kategorycznych – czym się różnią?

| Wykres | Co pokazuje? | Kiedy wybierać? | Główne ograniczenie |
|---|---|---|---|
| `barplot` | Estymator wartości liczbowej (domyślnie średnia) + przedział niepewności | Porównanie średniej/mediany lub innej statystyki między kategoriami | Nie pokazuje rozkładu ani pojedynczych rekordów; sprawdź `estimator` i `errorbar` |
| `countplot` | Liczbę rekordów w kategorii (lub procent) | Gdy pytanie dotyczy liczebności | Nie pokazuje wartości zmiennej ciągłej |
| `boxplot` | Medianę, IQR i obserwacje poza wąsami | Szybkie porównanie rozrzutu wielu grup | Nie pokazuje szczegółowego kształtu rozkładu |
| `violinplot` | Gęstość rozkładu (KDE), często wraz z kwartylami | Gdy ważny jest kształt rozkładu lub wielomodalność | Kształt zależy od wygładzania KDE i liczebności |
| `boxenplot` | Więcej poziomów kwantyli niż boxplot | Duże próby i porównanie ogonów rozkładu | Mniej intuicyjny; przy małej próbie zwykle nie daje przewagi |
| `stripplot` | Każdą obserwację jako punkt z losowym jitterem | Małe/średnie próby, gdy chcesz zobaczyć surowe dane | Punkty mogą na siebie nachodzić |
| `swarmplot` | Każdą obserwację, z rozmieszczeniem ograniczającym nakładanie | Małe próby, gdy ważna jest widoczność punktów | Wolniejszy; przy wielu punktach zgłasza problemy z rozmieszczeniem |
| `pointplot` | Estymator (domyślnie średnią) i przedział niepewności | Porównanie estymat między kategoriami, zwłaszcza dla układu `hue` | Łączące linie są wizualnym przewodnikiem, nie testem interakcji |
| `catplot` | Żaden nowy kształt – opakowuje powyższe typy | Gdy potrzebujesz paneli `row` / `col` bez ręcznego tworzenia subplots | Figure-level: nie przyjmuje `ax=` i tworzy własną figurę |

**Najkrótsza reguła wyboru:** liczebności → `countplot`; estymata → `barplot` lub `pointplot`; rozkład → `boxplot` / `violinplot` / `boxenplot`; surowe obserwacje → `stripplot` / `swarmplot`; panele → `catplot`.

`barplot` i `countplot` odpowiadają na inne pytania: pierwszy agreguje zmienną liczbową, drugi zlicza wiersze. `catplot(kind="box")` nie tworzy innego wykresu niż `boxplot()` – daje ten sam boxplot w interfejsie figure-level.

| `edgecolor`, `linewidth` ⬜ | Obrys punktów |
| `marker` ⬜ | Znacznik stały |
| `legend` ⬜ | `"auto"`, `"brief"`, `"full"`, `False` |
| `palette`, `hue_order` ⬜ | |

````python
smp = df.sample(600, random_state=4)
fig, axes = plt.subplots(2, 2, figsize=(14, 11), constrained_layout=True)

# a) hue kategoryczne + style
sns.scatterplot(
    smp,
    x="unit_price",
    y="revenue",
    hue="segment",
    style="channel",
    palette="viridis",
    s=45,
    alpha=0.7,
    edgecolor="white",
    linewidth=0.4,
    ax=axes[0, 0],
)
axes[0, 0].set_title("hue + style")
sns.move_legend(axes[0, 0], "upper left", bbox_to_anchor=(1.01, 1))

# b) hue ciągłe + size
sns.scatterplot(
    smp,
    x="unit_price",
    y="revenue",
    hue="discount",
    size="units",
    palette="flare",
    sizes=(15, 200),
    alpha=0.7,
    ax=axes[0, 1],
)
sns.move_legend(axes[0, 1], "upper left", bbox_to_anchor=(1.01, 1), fontsize=8)
axes[0, 1].set_title("hue (ciągłe) + size")

# c) wyróżnienie podzbioru + adnotacje + linie
sns.scatterplot(smp, x="unit_price", y="revenue", color="#c7cdd6", s=25, ax=axes[1, 0])
top = smp.nlargest(5, "revenue")
axes[1, 0].scatter(
    top["unit_price"],
    top["revenue"],
    s=110,
    color="#d62728",
    edgecolor="black",
    zorder=3,
    label="Top 5",
)
for _, r in top.iterrows():
    axes[1, 0].annotate(
        r["order_id"],
        (r["unit_price"], r["revenue"]),
        xytext=(6, 6),
        textcoords="offset points",
        fontsize=8,
    )
axes[1, 0].axhline(
    smp["revenue"].quantile(0.95), color="black", linestyle="--", linewidth=1
)
axes[1, 0].text(
    smp["unit_price"].min(),
    smp["revenue"].quantile(0.95),
    " P95",
    va="bottom",
    fontsize=9,
)
axes[1, 0].legend(loc="upper left", bbox_to_anchor=(1.01, 1), borderaxespad=0)
axes[1, 0].set_title("Wyróżnione punkty, adnotacje, próg P95")

# d) ręczna linia regresji (polyfit) + r + pasmo
x, y = smp["unit_price"].to_numpy(), smp["revenue"].to_numpy()
slope, intercept = np.polyfit(x, y, 1)
r = np.corrcoef(x, y)[0, 1]
xs = np.linspace(x.min(), x.max(), 100)
resid_sd = np.std(y - (slope * x + intercept))
sns.scatterplot(
    smp, x="unit_price", y="revenue", color="#4c72b0", alpha=0.5, ax=axes[1, 1]
)
axes[1, 1].plot(
    xs,
    slope * xs + intercept,
    color="crimson",
    linewidth=2,
    label=f"y = {slope:.2f}x + {intercept:.1f}",
)
axes[1, 1].fill_between(
    xs,
    slope * xs + intercept - 2 * resid_sd,
    slope * xs + intercept + 2 * resid_sd,
    color="crimson",
    alpha=0.1,
    label="±2 σ reszt",
)
axes[1, 1].text(
    0.05,
    0.92,
    f"r = {r:.2f}",
    transform=axes[1, 1].transAxes,
    fontsize=12,
    bbox={"boxstyle": "round", "facecolor": "white", "alpha": 0.8},
)
axes[1, 1].legend(loc="upper left", bbox_to_anchor=(1.01, 1), borderaxespad=0)
axes[1, 1].set_title("Regresja liniowa ręcznie (np.polyfit)")
plt.show()
````

### 9.5. Radar Chart – porównanie profilu wielu wskaźników

Radar chart (wykres radarowy / pajęczynowy) porównuje kilka wymiarów jednocześnie. Każdy wymiar jest osobną osią wychodzącą ze wspólnego środka.

**Warunki i ograniczenia:**
- wskaźniki muszą być sprowadzone do porównywalnej skali (np. 0–100); nie mieszaj bezpośrednio PLN, procentów i liczby klientów;
- porównuj niewiele obiektów i umiarkowaną liczbę wymiarów;
- powierzchnia wielokąta może wzmacniać optycznie różnice; do precyzyjnego rankingu często lepszy jest wykres punktowy lub słupkowy.

Pierwszą wartość powtarza się na końcu, aby zamknąć obrys wielokąta.

### Korelacja – Pearson i Spearman

```python
num = df.select_dtypes(include="number")

corr_pearson = num.corr(method="pearson")
corr_spearman = num.corr(method="spearman")
```

- **Pearson** mierzy siłę i kierunek zależności liniowej.
- **Spearman** mierzy zależność monotoniczną na podstawie rang; jest mniej wrażliwy na skalę i nie wymaga liniowości.

Współczynnik korelacji nie opisuje wszystkich możliwych zależności. Zależność nieliniowa, ale niemonotoniczna, może dać niski wynik zarówno Pearsona, jak i Spearmana. Warto obejrzeć scatterplot.

Przy interpretacji:
- `+1` – idealna dodatnia zależność;
- `0` – brak wykrytej zależności danego typu, niekoniecznie brak jakiejkolwiek zależności;
- `-1` – idealna ujemna zależność.

Korelacja nie oznacza przyczynowości. Wysoka korelacja może wynikać m.in. ze wspólnego trendu, zmiennej trzeciej, duplikacji informacji lub konstrukcji zmiennych pochodnych.

| Parametr | Opis |
|---|---|
| `data` ✅ | Macierz 2D (DataFrame, najlepiej z etykietami) |
| `vmin`, `vmax` ⬜ | Zakres kolorów |
| `center` ⬜ | Wartość środkowa palety rozbieżnej (np. `0`) |
| `cmap` ⬜ | Colormapa (`"coolwarm"`, `"vlag"`, `"rocket"`, `"YlGnBu"`) |
| `annot` ⬜ | `True` lub macierz własnych etykiet |
| `fmt` ⬜ | Format liczb: `".2f"`, `",.0f"`, `".0%"`, `"d"` |
| `annot_kws` ⬜ | `{"size": 9, "weight": "bold"}` |
| `linewidths`, `linecolor` ⬜ | Siatka między komórkami |
| `cbar`, `cbar_kws` ⬜ | Pasek kolorów: `{"label": "r", "shrink": .8, "orientation": "horizontal"}` |
| `square` ⬜ | Kwadratowe komórki |
| `mask` ⬜ | Macierz bool, `True` = komórka ukryta (np. górny trójkąt) |
| `xticklabels`, `yticklabels` ⬜ | `"auto"`, `True`, `False`, krok (int) |
| `robust` ⬜ | Zakres kolorów z percentyli 2–98 (odporne na outliery) |

````python
fig, axes = plt.subplots(2, 2, figsize=(15, 12), constrained_layout=True)

# a) korelacje: dolny trójkąt, rozbieżna paleta wycentrowana na 0
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="vlag",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5,
    linecolor="white",
    cbar_kws={"shrink": 0.7, "label": "Pearson r"},
    annot_kws={"size": 9},
    ax=axes[0, 0],
)
axes[0, 0].set_title("Macierz korelacji (dolny trójkąt)")

# b) tabela przestawna, sekwencyjna paleta
sns.heatmap(
    pivot,
    annot=True,
    fmt=",.0f",
    cmap="YlGnBu",
    linewidths=0.5,
    cbar_kws={"label": "Śr. przychód"},
    ax=axes[0, 1],
)
axes[0, 1].set_title("Region × segment")

# c) kalendarz aktywności: liczba zamówień wg miesiąca i dnia tygodnia
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
cal = (
    df.assign(m=df["order_date"].dt.month)
    .pivot_table(index="weekday", columns="m", values="order_id", aggfunc="count")
    .reindex(days)
)
sns.heatmap(
    cal,
    cmap="rocket_r",
    annot=True,
    fmt=".0f",
    linewidths=1,
    linecolor="white",
    ax=axes[1, 0],
)
axes[1, 0].set_title("Zamówienia: dzień tygodnia × miesiąc")
axes[1, 0].set_xlabel("Miesiąc")
axes[1, 0].set_ylabel("")

# d) obramowanie wybranej komórki i etykiety własne (annot jako macierz)
share = pivot.div(pivot.sum(axis=1), axis=0)
labels = (
    pivot.round(0).astype(int).astype(str)
    + "\n("
    + (share * 100).round(0).astype(int).astype(str)
    + "%)"
)
sns.heatmap(share, annot=labels, fmt="", cmap="crest", cbar=False, ax=axes[1, 1])
r_i, c_i = np.unravel_index(
    np.argmax(share.to_numpy()), share.shape
)  # komórka z maksimum
axes[1, 1].add_patch(
    Rectangle((c_i, r_i), 1, 1, fill=False, edgecolor="red", linewidth=3)
)
axes[1, 1].set_title("Etykiety własne + obramowana komórka max")
plt.show()
````

### 10.2. Clustermap: `sns.clustermap`

**Kiedy:** grupowanie hierarchiczne wierszy/kolumn. Odkrywanie grup skorelowanych zmiennych, segmentów klientów, podobnych obiektów. Figure-level.
**Interpretacja:** wiersze/kolumny są **przestawione** tak, by podobne leżały obok siebie; dendrogram pokazuje kolejność łączenia (im niżej połączenie, tym większe podobieństwo). Bloki o podobnym kolorze = klastry.

#### Wersja podstawowa (wymagane: `data`)

````python
sns.clustermap(corr, cmap="coolwarm", center=0)
plt.show()
````

| Parametr | Opis |
|---|---|
| `data` ✅ | Macierz |
| `method` ⬜ | Metoda łączenia: `"average"`, `"ward"`, `"complete"`, `"single"` (`"ward"` wymaga metryki euklidesowej) |
| `metric` ⬜ | Odległość: `"euclidean"`, `"correlation"`, `"cosine"`… |
| `standard_scale` / `z_score` ⬜ | Skalowanie `0` (wiersze) lub `1` (kolumny); **konieczne, gdy kolumny mają różne jednostki** |
| `row_cluster`, `col_cluster` ⬜ | Wyłączenie klastrowania danej osi |
| `row_colors`, `col_colors` ⬜ | Paski z kategorią (np. segment) obok wierszy/kolumn |
| `figsize`, `dendrogram_ratio`, `cbar_pos`, `colors_ratio` ⬜ | Układ |
| `cmap`, `center`, `vmin`, `vmax`, `annot`, `fmt`, `linewidths` ⬜ | Jak w `heatmap` |

````python
# a) klastrowanie zmiennych po korelacji
g = sns.clustermap(
    corr,
    method="average",
    metric="euclidean",
    cmap="vlag",
    center=0,
    vmin=-1,
    vmax=1,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    figsize=(8, 8),
    dendrogram_ratio=(0.15, 0.15),
    cbar_pos=(0.02, 0.8, 0.03, 0.15),
)
plt.show()

# b) klastrowanie obserwacji (wiersze) ze skalowaniem kolumn + pasek segmentu
num_cols = [
    "customer_age",
    "income",
    "units",
    "unit_price",
    "discount",
    "revenue",
    "delivery_days",
    "satisfaction",
]
sub = df.dropna(subset=num_cols).sample(80, random_state=5)
lut = dict(zip(df["segment"].cat.categories, sns.color_palette("Set2", 3)))
row_colors = sub["segment"].astype(str).map(lambda s: lut[s])
row_colors.name = "segment"

g = sns.clustermap(
    sub[num_cols],
    z_score=1,
    method="ward",
    cmap="vlag",
    center=0,
    robust=True,
    row_colors=row_colors.to_numpy(),
    yticklabels=False,
    figsize=(9, 8),
    dendrogram_ratio=(0.12, 0.15),
    cbar_kws={"label": "z-score"},
)
g.ax_heatmap.set_xlabel("")
plt.show()
````

### 10.3. Calendar Heatmap – aktywność w kolejnych dniach

Calendar heatmap układa dni tygodnia w wierszach, a tygodnie w kolumnach. Kolor komórki pokazuje liczbę zdarzeń, np. zamówień. Dzięki temu łatwiej zauważyć sezonowość tygodniową i dni o podwyższonej aktywności.

W przykładzie używam tygodni ISO. Uzupełniam również dni bez zamówień wartością zero, a nie pozostawiam ich jako brakujących komórek.

````python
daily_orders = df.groupby("order_date").size().rename("orders")
iso_year = int(df["order_date"].dt.isocalendar().year.mode().iloc[0])
jan4 = pd.Timestamp(f"{iso_year}-01-04")
start_week = jan4 - pd.Timedelta(days=jan4.dayofweek)
next_jan4 = pd.Timestamp(f"{iso_year + 1}-01-04")
end_week = next_jan4 - pd.Timedelta(days=next_jan4.dayofweek) - pd.Timedelta(days=1)
calendar_dates = pd.date_range(start_week, end_week, freq="D")
calendar_series = daily_orders.reindex(calendar_dates, fill_value=0)
calendar_df = calendar_series.rename_axis("date").reset_index()
calendar_df["weekday"] = calendar_df["date"].dt.day_of_week
calendar_df["week"] = calendar_df["date"].dt.isocalendar().week.astype(int).to_numpy()
calendar = calendar_df.pivot(index="weekday", columns="week", values="orders").reindex(
    index=range(7)
)

fig, ax = plt.subplots(figsize=(13, 4.5), constrained_layout=True)
sns.heatmap(
    calendar,
    cmap="YlGnBu",
    linewidths=0.3,
    linecolor="white",
    cbar_kws={"label": "Liczba zamówień"},
    ax=ax,
)
ax.set(
    title=f"Calendar heatmap – zamówienia, rok ISO {iso_year}",
    xlabel="Tydzień ISO",
    ylabel="Dzień tygodnia",
    yticks=np.arange(7) + 0.5,
    yticklabels=["Pon", "Wt", "Śr", "Czw", "Pt", "Sob", "Niedz"],
)
plt.show()
````

## 12. Układy złożone: FacetGrid, dual axis, dashboard

### 12.1. `FacetGrid`: ten sam wykres w wielu panelach

**Pomysł:** `FacetGrid` najpierw tworzy siatkę pustych osi. Następnie `map_dataframe()` uruchamia wskazaną funkcję osobno dla podzbioru danych przypisanego do każdego panelu.

- `col="channel"` – osobny panel dla każdego kanału;
- `row="segment"` – osobny wiersz dla każdego segmentu;
- `row` i `col` razem – panel dla każdej kombinacji, np. segment × kanał;
- `map_dataframe(sns.scatterplot, x="unit_price", y="revenue")` – nazwy kolumn przekazujesz jako argumenty, a Seaborn automatycznie dostaje właściwy podzbiór w `data=`.

`FacetGrid` jest elastycznym narzędziem do własnych paneli. Gdy wystarcza standardowy wykres, krótsze API to `relplot()` (scatter/line), `catplot()` (kategoryczne) lub `displot()` (rozkłady).

````python
# Każdy panel pokazuje ten sam scatterplot, ale dla innego kanału.
g = sns.FacetGrid(
    data=df, col="channel", col_wrap=3, height=3.0, aspect=1.1, sharex=True, sharey=True
)
g.map_dataframe(
    sns.scatterplot, x="unit_price", y="revenue", alpha=0.4, s=18, color="#4c72b0"
)
g.set_axis_labels("Cena jednostkowa", "Przychód")
g.set_titles("Kanał: {col_name}")
g.tight_layout()
plt.show()
````

| Parametr `FacetGrid` | Opis |
|---|---|
| `data` ✅ | DataFrame, który zostanie podzielony na podzbiory |
| `row`, `col` ⬜ | Kolumny dzielące dane na wiersze i kolumny paneli |
| `col_wrap` ⬜ | Zawijanie paneli w kolumnach (działa z `col`, bez `row`) |
| `height`, `aspect` ⬜ | Wysokość panelu i proporcja szerokość/wysokość |
| `sharex`, `sharey` ⬜ | Wspólne zakresy osi; ułatwiają uczciwe porównania |

Najważniejsze metody:

- `g.map(funkcja, "kolumna_x", "kolumna_y")` – przekazuje do funkcji same serie wartości; przydatne z funkcjami Matplotlib.

### 13.1. Linie

| Cel | Matplotlib (`ax.`) | Seaborn / uwagi |
|---|---|---|
| Pozioma / pionowa linia | `axhline(y, xmin, xmax)`, `axvline(x, ymin, ymax)` | `g.refline(x=, y=)` dla grid-ów |

#### Praktyczne przykłady: linie poziome, pionowe i ukośne

`axhline` i `axvline` przyjmują pozycję w jednostkach danych, natomiast `xmin`, `xmax`, `ymin`, `ymax` określają zakres w ułamku długości osi (od 0 do 1). Jeśli potrzebujesz odcinka o końcach wyrażonych w danych, użyj `hlines` lub `vlines`.

| Potrzeba | Przykład | Zastosowanie |
|---|---|---|
| Pozioma linia na całej szerokości | `ax.axhline(y=100, ...)` | cel, limit, średnia |
| Pionowa linia na całej wysokości | `ax.axvline(x=10, ...)` | data zdarzenia, próg |
| Odcinek poziomy / pionowy | `ax.hlines(...)`, `ax.vlines(...)` | przedział lub zakres |
| Linia z nachyleniem | `ax.axline((x0, y0), slope=...)` | linia trendu / odniesienia |
| Dowolna łamana | `ax.plot(x, y, ...)` | własne współrzędne |

`ax.axline` rysuje prostą nieskończoną. Dla linii ograniczonej do konkretnego zakresu użyj `ax.plot`.


````python
# Linie referencyjne na jednym wykresie
monthly = df.assign(month=df["order_date"].dt.to_period("M").dt.to_timestamp())
monthly_sales = monthly.groupby("month", as_index=False)["revenue"].sum()
avg_sales = monthly_sales["revenue"].mean()

fig, ax = plt.subplots(figsize=(10, 4.5), constrained_layout=True)
ax.plot(
    monthly_sales["month"],
    monthly_sales["revenue"],
    marker="o",
    ms=3,
    label="Przychód miesięczny",
)
ax.axhline(
    avg_sales,
    color="#d62728",
    linestyle="--",
    linewidth=1.6,
    label=f"Średnia: {avg_sales:,.0f}",
)
ax.axvline(
    monthly_sales["month"].median(),
    color="#2ca02c",
    linestyle=":",
    linewidth=1.6,
    label="Punkt środkowy okresu",
)
ax.set(
    title="Linie referencyjne: średnia i punkt w czasie",
    xlabel="Miesiąc",
    ylabel="Przychód",
)
ax.grid(axis="y", alpha=0.25)
ax.legend()
plt.show()

# Odcinki ograniczone do zakresu danych
fig, ax = plt.subplots(figsize=(8, 3.5), constrained_layout=True)
ax.plot([1, 2, 3, 4], [12, 16, 14, 19], marker="o")
ax.hlines(y=15, xmin=1, xmax=4, colors="#d62728", linestyles="--", label="Próg = 15")
ax.vlines(
    x=2.5, ymin=10, ymax=20, colors="#2ca02c", linestyles=":", label="Granica okresu"
)
ax.set(xlabel="Okres", ylabel="Wartość", title="Odcinki poziome i pionowe")
ax.legend()
ax.grid(alpha=0.2)
plt.show()
````

#### Linia ukośna 45° – linia równości `y = x`

Użyj jej, gdy porównujesz dwie wartości wyrażone w **tych samych jednostkach**, np. wartość rzeczywistą i przewidywaną.

- **Punkty na linii** – wartość X i Y są równe (idealna zgodność).
- **Punkty nad linią** – wartość na osi Y jest większa od X.
- **Punkty pod linią** – wartość na osi Y jest mniejsza od X.

Są dwa sposoby narysowania:

- `ax.plot([low, high], [low, high])` – odcinek ograniczony do podanego zakresu.
- `ax.axline((0, 0), slope=1)` – nieskończona prosta przechodząca przez punkt `(0, 0)` o nachyleniu `dy/dx = 1`.

**Dlaczego sama linia `y = x` może nie wyglądać na 45°?** Kąt widoczny na ekranie zależy również od zakresów oraz proporcji osi. Dla poprawnej wizualizacji ustaw takie same granice X i Y i `ax.set_aspect("equal", adjustable="box")`. Oznacza to, że ta sama odległość w danych na osi X i Y ma taką samą długość na ekranie. Nie wymuszaj linii 45° dla wielkości o różnych jednostkach – wtedy linia równości nie ma sensu.

````python
# Przykład: wartości rzeczywiste vs przewidywane – linia równości y = x
rng = np.random.default_rng(12)
y_true = np.linspace(20, 120, 80)
y_pred = y_true * 0.94 + rng.normal(0, 8, size=y_true.size)

# Wspólny zakres dla obu osi wyznaczamy z łącznych danych.
low = min(y_true.min(), y_pred.min())
high = max(y_true.max(), y_pred.max())

fig, ax = plt.subplots(figsize=(6.5, 6.5), constrained_layout=True)
ax.scatter(y_true, y_pred, alpha=0.65, s=35, edgecolor="white", linewidth=0.4)
ax.plot(
    [low, high],
    [low, high],
    color="#d62728",
    linestyle="--",
    linewidth=1.8,
    label="Idealna zgodność: y = x",
)
# Alternatywa: ax.axline((0, 0), slope=1, color="gray", linestyle=":")
ax.set_xlim(low, high)
ax.set_ylim(low, high)
ax.set_aspect("equal", adjustable="box")
ax.set(
    xlabel="Wartość rzeczywista (X)",
    ylabel="Wartość przewidywana (Y)",
    title="Porównanie wartości rzeczywistych i przewidywanych",
)
ax.grid(alpha=0.25)
ax.legend()
plt.show()
````

### 13.5. Etykiety danych: `bar_label`, `text`, `annotate`

Dobierz metodę do obiektu, który opisujesz:

- `ax.bar_label(container)` – najszybsza metoda do podpisywania słupków; `padding` ustala odstęp od końca, a `label_type="center"` umieszcza etykietę wewnątrz słupka.
- `ax.text(x, y, tekst)` – dowolny tekst w pozycji wyrażonej współrzędnymi danych lub osi.
- `ax.annotate(...)` – tekst powiązany z konkretnym punktem; pozwala dodać strzałkę (`arrowprops`) i przesunięcie (`xytext`).

Uważaj na liczbę etykiet: podpisuj tylko wartości, które faktycznie pomagają w odczycie. Przy wielu kategoriach etykiety mogą być mniej czytelne niż oś i legenda.

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.6), constrained_layout=True)

# a) bar_label – automatyczne etykiety nad słupkami
region_sum = df.groupby("region")["revenue"].sum().sort_values(ascending=False)
container = axes[0].bar(region_sum.index, region_sum.values, color="#4c72b0")
axes[0].bar_label(
    container,
    labels=[f"{v / 1_000:.0f}k" for v in region_sum.values],
    padding=3,
    fontsize=8,
)
axes[0].set(title="bar_label()", ylabel="Przychód")
axes[0].tick_params(axis="x", rotation=25)

# b) text – etykieta bezpośrednio przy punkcie
monthly = df.groupby("month", as_index=False)["revenue"].sum()
axes[1].plot(monthly["month"], monthly["revenue"], marker="o", ms=3)
max_row = monthly.loc[monthly["revenue"].idxmax()]
axes[1].text(
    max_row["month"], max_row["revenue"], "  maksimum", va="bottom", fontsize=9
)
axes[1].set(title="text()", ylabel="Przychód")
axes[1].tick_params(axis="x", rotation=35)

# c) annotate – punkt i opis połączone strzałką
axes[2].scatter(df["unit_price"], df["revenue"], alpha=0.25, s=15)
point = df.loc[df["revenue"].idxmax()]
axes[2].annotate(
    f"Najwyższy przychód: {point['revenue']:,.0f}",
    xy=(point["unit_price"], point["revenue"]),
    xytext=(15, -25),
    textcoords="offset points",
    arrowprops={"arrowstyle": "->", "color": "black", "lw": 1.2},
    bbox={"boxstyle": "round,pad=0.3", "fc": "white", "ec": "gray", "alpha": 0.9},
    fontsize=8,
)
axes[2].set(title="annotate()", xlabel="Cena", ylabel="Przychód")
plt.show()
````

### 14.3. Typowe pułapki

1. **Figure-level vs axes-level:** `sns.relplot(..., ax=ax)` nie zadziała. Do własnych subplots użyj `scatterplot`/`lineplot`; `FacetGrid` zwraca `g.axes`, `g.figure`.
2. **Seaborn ≥ 0.13:** przekazanie `palette` bez `hue` generuje ostrzeżenie (`FutureWarning`). Przypisz `hue=x` i `legend=False`.
3. **`set_xticklabels` bez `set_xticks`** (lub `FixedLocator`) daje przesunięte etykiety.
4. **Mieszanie `tight_layout` i `constrained_layout`:** wybierz jedno.
5. **Zmienne kategoryczne jako liczby** (`satisfaction` 1–5) seaborn traktuje jako ciągłe. Dla `catplot`/`boxplot` po osi X zadziała, ale dla `hue` ustaw `.astype("category")` albo `palette` dyskretną.
6. **Kolejność kategorii:** `pd.Categorical(..., ordered=True)` lub `order=` (np. dni tygodnia, miesiące).
7. **Oś Y barplot od zera.** Skracanie osi słupków zniekształca porównania.
8. **Ucięte pasma w `lineplot`:** dla jednej obserwacji na punkt X nie ma przedziału ufności, a `errorbar` nic nie zmieni.
9. **KDE poza zakresem danych:** ustaw `cut=0` lub `clip=(min, max)` (np. wartości nie mogą być ujemne).
10. **Daty:** upewnij się, że kolumna jest `datetime64`, a nie `object`/`str`.
11. **Rozmiar:** `figsize` w matplotlib (cale), a `height`/`aspect` w figure-level; `dpi` wpływa na rozmiar wyświetlania w notebooku.

## 15. Ściągawka: jaki wykres do jakiego pytania

| Pytanie analityczne | Wykres | Uwagi |
|---|---|---|
| Jak zmienia się wartość w czasie? | `lineplot`, `ax.plot` | Agreguj do właściwej częstotliwości; nie łącz kategorii nominalnych linią |
| Jak różnią się kategorie pod względem miary? | `barplot`, `bar`, lollipop, dot plot | Słupki zaczynaj od zera; lollipop i dot plot pokazują położenie punktów |
| Jak porównać dwie grupy / okresy w każdej kategorii? | Dumbbell chart | Dwa punkty i odcinek; dobre do porównania dwóch wartości |
| Jak znaleźć kategorie generujące największą część wyniku? | Pareto | Słupki malejąco + skumulowany udział; 80/20 nie jest prawem |
| Jak składniki budują wartość końcową? | Waterfall | Oddziel wartość początkową, zmiany i końcową |
| Jak wygląda rozkład jednej zmiennej? | `histplot`, `kdeplot`, `ecdfplot` | Histogram zależy od binów; KDE od bandwidth; ECDF nie wymaga wygładzania |
| Jak różnią się rozkłady grup? | `boxplot`, `violinplot`, `boxenplot` | Box = kwartyle/mediana; violin = KDE; boxen = więcej kwantyli |
| Jakie są liczebności kategorii? | `countplot` | Zlicza rekordy; nie agreguje miary liczbowej |
| Jak porównać estymaty między kategoriami? | `barplot`, `pointplot` | `barplot` używa słupków; `pointplot` punktów i zwykle linii |
| Jak zobaczyć pojedyncze obserwacje w grupach? | `stripplot`, `swarmplot` | Dla małych/średnich prób; swarm może być kosztowny |
| Jak pokazać X, Y i dodatkową miarę liczbową? | Bubble chart (`scatterplot(size=...)`) | Rozmiar punktu koduje dodatkową zmienną |
| Jaki jest związek dwóch zmiennych? | `scatterplot`, `regplot` | Korelacja nie oznacza przyczynowości |
| Co zrobić przy bardzo dużej liczbie punktów? | `hexbin`, histogram 2D | Ogranicz overplotting; kolor może oznaczać liczność lub agregat |
| Jak sprawdzić zgodność wartości rzeczywistych i przewidywanych? | Scatter + linia `y = x` | Te same jednostki, jednakowe zakresy i `aspect="equal"` |
| Szybki przegląd wielu zmiennych liczbowych | `pairplot`, `pd.plotting.scatter_matrix` | Pairplot ma więcej opcji; Pandas scatter matrix jest prostsza |
| Jak porównać profil kilku obiektów w wielu KPI? | Radar chart | Normalizuj skale; ogranicz liczbę osi i serii |
| Jakie zmienne są skorelowane? | `heatmap` | Dobierz Pearson/Spearman do pytania; korelacja nie opisuje każdej zależności |
| Jak pogrupować podobne zmienne / obserwacje? | `clustermap` | Standaryzacja wpływa na odległości i wynik klastrowania |
| Kiedy występuje aktywność w ciągu roku? | Calendar heatmap | Kolumny = tygodnie, wiersze = dni tygodnia; uzupełnij dni bez zdarzeń zerami |
| Jak porównać rozkłady wielu grup? | Ridge plot | Przy wielu grupach; wspólna oś X ułatwia porównanie |
| Ten sam wykres dla wielu kategorii | `FacetGrid`, `relplot`, `catplot`, `displot` | Small multiples; `catplot(kind="box")` to nadal boxplot, tylko figure-level |
| Dwie miary w różnych jednostkach | `twinx` | Stosuj oszczędnie; dobór skal może sugerować nieistniejące podobieństwa |

**Ważne rozróżnienia:** `barplot` agreguje zmienną liczbową, `countplot` zlicza rekordy. `catplot` i `relplot` są interfejsami figure-level, a nie pojedynczymi typami geometrii. `pairplot` i `scatter_matrix` prezentują analogiczny układ par zmiennych, ale mają różne możliwości i API.
