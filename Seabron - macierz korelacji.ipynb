```python
def corr_matrix(df):
    metoda = "pearson"  # "pearson" | "spearman" | "kendall"
    corr = df.select_dtypes(include="number").corr(method=metoda)

    # Z przekątną ukrytą pierwszy wiersz i ostatnia kolumna są w całości puste. Jeśli Ci to przeszkadza, utnij je i zostaw k=1
    corr = corr.iloc[1:, :-1]
    # mask: True = komórka ukryta. k=0 ukrywa górny trójkąt razem z przekątną (duplikat, a przekątna zawsze wynosi 1), k=1 ukryłby tylko trójkąt i zostawił przekątną.
    mask = np.triu(np.ones(corr.shape, dtype=bool), k=0)
    # mask |= (corr.abs() < 0.3).to_numpy()   # opcjonalnie: ukryj słabe korelacje (|r| < 0.3)

    fig, ax = plt.subplots(figsize=(8, 6))

    sns.heatmap(
        corr,
        mask=mask,  # patrz wyżej
        annot=True,  # wartości w komórkach; przy >15-20 zmiennych ustaw False
        fmt=f".{2}f",  # format liczb; .2f standard, .1f dla dużych macierzy
        annot_kws={"size": 9},  # czcionka wartości; zmniejsz przy gęstej macierzy
        cmap="RdBu_r",  # diverging: kolor mówi o znaku. Alternatywy: "coolwarm", "vlag". Unikaj sekwencyjnych (viridis), bo gubią znak
        center=0,  # zero korelacji = biały; nie zmieniaj
        vmin=-1,
        vmax=1,  # stała skala daje porównywalność wykresów; zawężaj tylko świadomie
        linewidths=0.5,  # siatka między komórkami; 0 przy 50+ zmiennych
        square=True,  # kwadratowe komórki; False przy bardzo długich nazwach kolumn
        cbar_kws={
            "shrink": 0.8,  # wysokość paska względem wykresu (przy square=True bywa za wysoki)
            "label": f"Współczynnik korelacji ({metoda.capitalize()})",
            "ticks": [-1, -0.5, 0, 0.5, 1],  # opcjonalnie: wyraźne znaczniki
        },
        ax=ax,
    )

    ax.tick_params(labelsize=10)
    plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")
    plt.tight_layout()

    return plt.show()
```
